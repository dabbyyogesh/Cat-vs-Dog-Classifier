# importing librayies

```python
import tensorflow as tf
from tensorflow.keras import layers, models # type: ignore
import numpy as np
import matplotlib.pyplot as plt
```

# using generator function to load the data in baches

```python
# Define the paths based on your download location
train_dir = '/kaggle/input/dogsvscats/train'
test_dir = '/kaggle/input/dogsvscats/test'

print("Loading training data...")
# using generator function
train_dataset = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    image_size = (150,150),
    batch_size = 32
)
print("loading testing data...")
test_dataset = tf.keras.utils.image_dataset_from_directory(
    test_dir,
    image_size = (150,150),
    batch_size = 32
)
```

# Normalization

```python
def process(image,label):
  image = tf.cast(image/255. , tf.float32)
  return image,label

train_dataset = train_dataset.map(process)
test_dataset = test_dataset.map(process)
```

# Model creation (CNN)

```python
model = models.Sequential([

    layers.Input(shape = (150,150,3)),

    # Data Augmentation
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),

    #first convulation layer
    layers.Conv2D(filters= 32 , kernel_size= (3,3) , activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D(pool_size=(2,2)),

    #second convualtion layer
    layers.Conv2D(filters= 64 ,kernel_size= (3,3) , activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D(pool_size=(2,2)),

    #third convulation layer
    layers.Conv2D(filters= 128 , kernel_size=(3,3) , activation= 'relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D(pool_size=(2,2)),

    # flatten the 2d array to 1d array
    layers.Flatten(),

    #dense layer for final decision
    layers.Dense(units=128 , activation= 'relu'),
    layers.Dropout(0.5),
    layers.Dense(units=64 , activation= 'relu'),
    layers.Dropout(0.1),

    #output layer
    layers.Dense(units=1 , activation='sigmoid' )
])

model.summary()
```

# Compiling

```python
model.compile(optimizer='adam',
              loss = 'binary_crossentropy',
              metrics = ['accuracy'])
print("starting training on the GPU...")

history = model.fit(
    train_dataset,
    validation_data=test_dataset,
    epochs=10
)
```

```python
import matplotlib.pyplot as plt

plt.plot(history.history['accuracy'],color = 'r' , label = 'train')
plt.plot(history.history['val_accuracy'],color = 'b', label  = ' validation')
plt.legend()
plt.show()
```

```python
plt.plot(history.history['loss'],color = 'r' , label = 'train')
plt.plot(history.history['val_loss'],color = 'b', label  = ' validation')
plt.legend()
plt.show()
```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

# Create a list of your uploaded image filenames
image_files = ['cat.jpg', 'dog.jpg']

# Set up a plot to show both images side-by-side
plt.figure(figsize=(10, 5))

for i, img_path in enumerate(image_files):
    # 1. Load and resize the image
    img = tf.keras.utils.load_img(img_path, target_size=(150, 150))

    # 2. Convert to array and scale pixels (0 to 1)
    img_array = tf.keras.utils.img_to_array(img)
    img_array = img_array / 255.0

    # 3. Add the batch dimension
    img_array = tf.expand_dims(img_array, 0)

    # 4. Make the prediction
    predictions = model.predict(img_array)
    score = predictions[0][0]

    # 5. Display the results
    plt.subplot(1, 2, i + 1)
    plt.imshow(img)
    plt.axis('off')

    if score > 0.5:
        plt.title(f"Prediction: DOG 🐶\n({score*100:.1f}% confident)")
    else:
        plt.title(f"Prediction: CAT 🐱\n({(1-score)*100:.1f}% confident)")

plt.tight_layout()
plt.show()
```
